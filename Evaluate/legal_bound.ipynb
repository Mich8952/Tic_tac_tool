```python
import pandas as pd
import numpy as np
from math import comb # choose function in python
import matplotlib.pyplot as plt
```

```python
def valid_moves_count(n):
    grid = n**2
    total = 0
    for x in range(grid+1):
        for o in range(grid+1):
            if x + o <= grid and (x == o or x == o + 1):
                total += comb(grid, x) * comb(grid - x, o)
    return total
```

```python
print(valid_moves_count(2))
```

```python
print(valid_moves_count(3))
```

```python
Ns = np.arange(2,15+1)
moves = [valid_moves_count(n) for n in Ns]
plt.rcParams.update({'font.size': 14})  
plt.clf()
plt.plot(Ns,moves, color = 'black')
plt.yscale('log')
plt.ylabel("Number of Valid States")
plt.xlabel('Board Dimension (n)')
plt.title("Growth of Valid States with Board Dimension (n)")
plt.show()
```
